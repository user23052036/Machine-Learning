Linear Regression:

**Y = wX + b**

Y --> Dependent Variable<br>
X --> Independent Variable

w --> weight
b --> bias

**Gradient Descent:**

Gradient Descent is an optimization algorithm used for minimizing the loss function in various machine learning algorithms. It is used for updating the parameters of the learning model.

w  =  w - α*dw
b  =  b - α*db

**Learning Rate:**

Learning rate is a tuning parameter in an optimization algorithm that determines the step size at each iteration while moving toward a minimum of a loss function.

## How Gradient Descent Works for Linear Regression

A concise, exam-ready explanation and derivation for gradient descent applied to simple linear regression.

---

### Notation & Model
- Training set: $\{(x_i, y_i)\}_{i=1}^n$  
- Prediction:  
  $$
  y_i^{pred} = m x_i + b
  $$
  where $m$ is the slope (weight) and $b$ is the intercept (bias).

---

### 1. Cost Function (Mean Squared Error)
We measure error with the Mean Squared Error (MSE):

$$
J(m, b) = \frac{1}{n} \sum_{i=1}^{n} \bigl(y_i - y_i^{pred}\bigr)^2
= \frac{1}{n} \sum_{i=1}^{n} \bigl(y_i - (m x_i + b)\bigr)^2
$$

The goal: find $m, b$ that minimize $J(m,b)$.

---

### 2. Compute Partial Derivatives (Gradients)

#### Partial derivative w.r.t. slope $m$
Start with
$$
D_m \equiv \frac{\partial J}{\partial m}
= \frac{\partial}{\partial m}\left(\frac{1}{n}\sum_{i=1}^n (y_i - (m x_i + b))^2\right)
$$

Pull constant outside and apply chain rule:
$$
D_m = \frac{1}{n}\sum_{i=1}^n 2\bigl(y_i - (m x_i + b)\bigr)\cdot\frac{\partial}{\partial m}\bigl(y_i - (m x_i + b)\bigr)
$$

Since $\frac{\partial}{\partial m}\bigl(y_i - (m x_i + b)\bigr) = -x_i$:
$$
D_m = \frac{1}{n}\sum_{i=1}^n 2\bigl(y_i - (m x_i + b)\bigr)(-x_i)
= -\frac{2}{n}\sum_{i=1}^n x_i\bigl(y_i - (m x_i + b)\bigr)
$$

Boxed form:
$$
\boxed{D_m = -\frac{2}{n}\sum_{i=1}^n x_i\,(y_i - y_i^{pred})}
$$

---

#### Partial derivative w.r.t. intercept $b$
Start with
$$
D_b \equiv \frac{\partial J}{\partial b}
= \frac{\partial}{\partial b}\left(\frac{1}{n}\sum_{i=1}^n (y_i - (m x_i + b))^2\right)
$$

Apply chain rule (and $\frac{\partial}{\partial b}(y_i - (m x_i + b)) = -1$):
$$
D_b = \frac{1}{n}\sum_{i=1}^n 2\bigl(y_i - (m x_i + b)\bigr)(-1)
= -\frac{2}{n}\sum_{i=1}^n \bigl(y_i - (m x_i + b)\bigr)
$$

Boxed form:
$$
\boxed{D_b = -\frac{2}{n}\sum_{i=1}^n (y_i - y_i^{pred})}
$$

---

### 3. Gradient Descent Update Rules
Using learning rate $\alpha$:

$$
m \leftarrow m - \alpha\,D_m
\qquad\text{and}\qquad
b \leftarrow b - \alpha\,D_b
$$

Substituting $D_m, D_b$ if you prefer the explicit update:

$$
m \leftarrow m + \alpha\left(\frac{2}{n}\sum_{i=1}^n x_i(y_i - y_i^{pred})\right)
$$

$$
b \leftarrow b + \alpha\left(\frac{2}{n}\sum_{i=1}^n (y_i - y_i^{pred})\right)
$$

(The negative in $D_m,D_b$ becomes a plus in the updates because we subtract the gradient.)

---

### 4. Algorithm (Batch Gradient Descent)
1. Initialize $m, b$ (e.g., small random values or zeros).  
2. Repeat until convergence:
   - Compute predictions $y_i^{pred} = m x_i + b$ for all $i$.
   - Compute gradients $D_m, D_b$ using formulas above.
   - Update parameters using the update rules.
3. Stop when change in $J(m,b)$ or parameters is below a small threshold, or after a max number of iterations.

---

### Practical notes, failure modes & fixes
- **Notation consistency:** Pick one notation and use it throughout ($m,b$ here). Mixing $b$ and $c$ or $w$ and $m$ is confusing and cost you marks in exams.
- **Learning rate $\alpha$**:
  - Too large ⇒ divergence / exploding loss.
  - Too small ⇒ extremely slow convergence.
  - Use learning-rate schedules or try values in $\{10^{-3},10^{-2},10^{-1}\}$ scaled by feature magnitude.
- **Feature scaling (standardization)**: strongly recommended if $x$ values have different ranges — speeds convergence and stabilizes choice of $\alpha$.
- **Batch vs. mini-batch vs. stochastic**:
  - Batch (above): uses full dataset each update — stable gradients but costly on large $n$.
  - Mini-batch: common practical compromise.
  - Stochastic: noisy updates, can escape shallow local issues.
- **Stopping criteria**: small change in $J$, small gradient norm, or patience-based early stopping on validation set.
- **Vectorized form**: use matrix operations for efficiency in code/real projects.
- **Common exam trap**: show chain-rule step — don’t skip it. Exams expect the derivative step explicitly.

---

### Final boxed summary
- Linear Model:
  $$
  Y = wX + b
  $$

- Cost:
  $$
  J(w,b)=\frac{1}{n}\sum_{i=1}^n \bigl(y_i - (w x_i + b)\bigr)^2
  $$

- Gradients:
  $$
  D_w = -\frac{2}{n}\sum_{i=1}^n x_i\bigl(y_i - y_i^{pred}\bigr),\qquad
  D_b = -\frac{2}{n}\sum_{i=1}^n \bigl(y_i - y_i^{pred}\bigr)
  $$

- Update:
  $$
  w \leftarrow w - \alpha D_w,\qquad
  b \leftarrow b - \alpha D_b
  $$

```python
# importing numpy library
import numpy as np
```

```python
# model parameters -> weight, bias
# hyperparameters -> learning rate, no of iterations

class Linear_Regression():

    def __init__(self, learning_rate, no_of_iterations):
        self.learning_rate = learning_rate
        self.no_of_iterations = no_of_iterations


    # fit the data into out model
    def fit(self, X, Y):
        # number of training examples and number of features

        self.m, self.n = X.shape # no of rows and columns(here n=1)

        self.w = np.zeros(self.n) # innitialize weight
        self.b = 0 # innitialized bias
        self.X = X
        self.Y = Y

        # implementing gradient descent
        for i in range(self.no_of_iterations):
            self.update_weights()
    

    def update_weights(self):
        Y_predicted = self.predict(self.X)

        dw = (-2/self.m)*((self.X.T).dot(self.Y - Y_predicted))
        db = (-2/self.m)*(self.Y - Y_predicted)

        
    
    def predict(self, X):
        return X.dot(self.w) + self.b
```
