```python
data = [101.1,102.4,103.3]
series = pd.Series(data, index=['a','b','c'])
print(series)
print(series.loc['a'])

series.loc['a'] = 199.1
print(series.iloc[0])
```

```python
fruits = { 'apples':[3,2,0,1] , 'oranges':[0,3,7,2] , "angur":[1,0,2,1] , 'bananas':[4,3,5,7] }
series = pd.Series(fruits)
df = pd.DataFrame(fruits)
print(df)
# we can use dict key as labels

series.loc['oranges'].append(32)
print(series)

print(series[series.apply(sum) < 15])
```

```python
company_info = {
    'name': ['sonu','rushi','loki','anisha'],
    'age': [12,11,13,44],
    'isAdult': [False, False, False, True]
}

df = pd.DataFrame(company_info, index= ['employee1','employee2','employee3','employee4'])

# adding a new column
df['job'] = ['student','wall-painter','mechanic','artist']

# adding a new rows
new_rows = pd.DataFrame([ {'name':'crabs', 'age':21, 'isAdult':True, 'job':'manager'},
                          {'name':'squid', 'age':17, 'isAdult':False, 'job':'Bae-Dancer'}],
                        index=['employee5','employee6'])
print(new_rows)
print('\n\n')
df = pd.concat([df,new_rows])
print(df)
```
