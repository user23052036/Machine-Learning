### Aggregate Function 


```python

# These aggregate function is applied to the whole dataframe
# print(df.mean(numeric_only=True))
# print(df.sum(numeric_only=True))
# print(df.min(numeric_only=True))
# print(df.max(numeric_only=True))
# print(df.count()) # count does not include null values

# single column
print(df["Height"].mean(numeric_only=True))
print(df["Height"].sum(numeric_only=True))
print(df["Height"].min(numeric_only=True))
print(df["Height"].max(numeric_only=True))
print(df["Height"].count()) # count does not include null values
```

```python
# groupby() splits the dataframe into groups based on a column.
# Here, rows are grouped according to the values in the "Type1" column.

group = df.groupby("Type1")

# No calculation happens yet.
# We must apply an aggregation operation like mean(), min(), max(), sum(), etc.

# Find the minimum Height value for each Type1 group
print(group["Height"].min())

# Example:
# Fire   -> minimum height among all Fire Pokémon
# Water  -> minimum height among all Water Pokémon
```

### Data Cleaning
The process of removing incorrect,irrelivent and incomplete data.
75% of work done in pandas is for data cleaning

```python
# 1. drop irrelivent columns
df = df.drop(columns= ["No","Legendary"])
```

```python
# Handle missing data (NaN values)

# dropna() removes rows containing missing values.

# If any row has NaN in the "Type2" column,
# the entire row will be removed.
df = df.dropna(subset=["Type2"])


# fillna() replaces missing values instead of deleting rows.

# Here, all NaN values in the "Type2" column are replaced with the string "None".
df = df.fillna({"Type2": "None"})
```

```python
# Fix inconsistent values in a column
# replace() changes specific values into new values.

# Here, values in the "Type1" column are converted
# into a consistent uppercase format.

df["Type1"] = df["Type1"].replace({
    "Grass": "GRASS",
    "Fire": "FIRE",
    "Water": "WATER"
})

# Example:
# Grass -> GRASS
# Fire  -> FIRE
# Water -> WATER
```

```python
# 4. standarized text
df["Type2"] = df["Type2"].str.lower()
df.index = df.index.str.lower()
df.head()
```

```python
# 5. fix datatypes
df["Legendary"] = df["Legendary"].astype(bool)
```

```python
# 6. removing duplicate entries
df = df.drop_duplicates()
print(df)
```

```python
print("Columns:", df.columns.tolist())
print("Index name:", df.index.name)
print("Index type:", type(df.index))
print(df.head())
```
