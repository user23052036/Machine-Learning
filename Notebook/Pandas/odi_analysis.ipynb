```python
import pandas as pd
```

```python
df = pd.read_csv('../../Data/odi-batting.csv')
df.head()
```

```python
# by default describe() only works on numeric columns 
# but it is good practice to only select the columns you want to work with

select_col = ['Runs','ScoreRate']
df[select_col].describe()
```

```python
df.shape
```

```python
df.info()
```

```python
# now to describe the categorical columns we chosse only the object columns
cat_col = ['Country','Player','Ground','Versus']
df[cat_col].astype('object').describe()

# on numerical columns describe() provides distribution analysis and on 
# categorical columns it provides frequency analysis.
```

```python
df['MatchDate'] = pd.to_datetime(df['MatchDate'], format='%d-%m-%Y')
df.dtypes # this is how we can check the data types of the columns in our dataframe
```

```python
df['MatchDate'].describe()
```
