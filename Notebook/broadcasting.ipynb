## broadcasting

```python
A = np.array([
#   Apple  Orange Guava   Cherry
    [56.0, 0.0,    4.4,   68.0], # carbs
    [1.2, 104.0,   52.0,   8.0], # protein
    [1.8, 135.0,   99.0,   0.9]  # Fat
])
A
```

```python
cal = A.sum(axis=0) # column-wise
cal
```

```python
cal = cal.reshape(4,1)
cal.shape
```

```python
cal
```

```python
A.shape
```

```python
# calculate the percentage of carbs, protein and fat in each fruit-type column-wise
percentage = 100*A.T/cal
percentage
```

```python
percentage = 100*A/cal.reshape(1,4)
percentage
```
