```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.metrics import accuracy_score

from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
dataset = pd.read_csv('../Data/heart.csv')
```

```python
dataset.head()
```

### preprocessing

```python
# be very careful as standarization is not applicable for categorical datas.. 
# for example here sex
```

```python
dataset.shape
```

```python
dataset.isnull().sum()
```

```python
dataset.groupby('target').size()

# 1 --> defective heart
# 0 --> normal heart
```

```python
X = dataset.iloc[:, 0:13]
# dataset.drop(column='target', axis=1)
X
```

```python
Y = dataset.iloc[:, -1]
Y
```

### Train test split

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,
                                                 test_size=0.15,
                                                 random_state=2,
                                                 stratify=Y)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

### Comparing the performance of the models

```python
# list of models
models = [LogisticRegression(max_iter=1000), SVC(kernel='linear'), KNeighborsClassifier(), RandomForestClassifier()]
```

```python
def compare_models_train_test():

  for model in models:

    # training the model
    model.fit(X_train, Y_train)
    
    # evaluating the model
    test_data_prediction = model.predict(X_test)
    accuracy = accuracy_score(Y_test, test_data_prediction)

    print('Accuracy score of the ', model, ' = ', accuracy)
```

```python
compare_models_train_test()
```

### Cross Validation
<img src='../images/cross_val_image.png'>

Logistic Regression

```python
cv_score_lr = cross_val_score(LogisticRegression(max_iter=1000),X,Y,cv=5)
# if we mention the value of cv instead of using kfold we use stratified k-fold method
print(cv_score_lr)

mean_accuracy_lr = sum(cv_score_lr)/len(cv_score_lr)
mean_accuracy_lr *= 100
mean_accuracy_lr = round(mean_accuracy_lr,2)
print(f"{mean_accuracy_lr}%")
```

SVC(support Vector Classifier)

```python
cv_score_sv = cross_val_score(SVC(kernel='linear'),X,Y,cv=5)
# if we mention the value of cv instead of using kfold we use stratified k-fold method
print(cv_score_sv)

mean_accuracy_sv = sum(cv_score_sv)/len(cv_score_sv)
mean_accuracy_sv *= 100
mean_accuracy_sv = round(mean_accuracy_sv,2)
print(f"{mean_accuracy_sv}%")
```

KNN(k nearest neighbour)

```python
cv_score_knn = cross_val_score(KNeighborsClassifier(),X,Y,cv=5)
# if we mention the value of cv instead of using kfold we use stratified k-fold method
print(cv_score_knn)

mean_accuracy_knn = sum(cv_score_knn)/len(cv_score_knn)
mean_accuracy_knn *= 100
mean_accuracy_knn = round(mean_accuracy_knn,2)
print(f"{mean_accuracy_knn}%")
```

Random Forest Classifier

```python
cv_score_rfc = cross_val_score(RandomForestClassifier(),X,Y,cv=5)
# if we mention the value of cv instead of using kfold we use stratified k-fold method
print(cv_score_rfc)

mean_accuracy_rfc = sum(cv_score_rfc)/len(cv_score_rfc)
mean_accuracy_rfc *= 100
mean_accuracy_rfc = round(mean_accuracy_rfc,2)
print(f"{mean_accuracy_rfc}%")
```
